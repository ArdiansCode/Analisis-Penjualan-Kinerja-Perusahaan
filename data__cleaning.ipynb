# **Import Library**

```python
import pandas as pd
```

# **Muat Dataset**

```python
df = pd.read_csv("/content/raw_sales_data_messy.csv")

df.head()
```

```python
df.info()
```

### **mengubah Dtype datetime ke format datetime**

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

# **Pembersihan Data**

### **Cek Mising value**

```python
df.isnull().sum()
```

### **Membersihkan kolom city dengan mengisi missing value dengan modus**

```python
modus_city = df["City"].mode()[0]

df["City"] = df['City'].fillna(modus_city)
```

### **Membersihkan Missing value kolom Item_Price mengunakan nilai mean item price bergasarkan kategory**

```python
df['Item_Price'] = df['Item_Price'].fillna(df.groupby('Product_Category')['Item_Price'].transform('mean'))
```

### **Membersihkan Missing value kolom Quantityt mengunakan nilai mean item price bergasarkan kategory**

```python
df['Quantity'] = df['Quantity'].fillna(df.groupby('Product_Category')['Quantity'].transform('mean').round())
```

```python
df.isnull().sum()
```

# **Feature Engineering**

### **Membuat kolom baru yang berisi nilai jual setelah diskon**

```python
df['Net_Item_Price'] = df['Item_Price'] * (1 - df['Discount_Percent'] / 100)
```

### **Membuat kolom baru yang berisi nilai total pendapatan transaksi**

```python
df['Gross_Revenue'] = df['Item_Price'] * df['Quantity']
```

### **Membuat kolom baru yang berisi nilai total pendapatan jika dihitung dengan ongkir**

```python
df['Total_Transaction_Value'] = df['Gross_Revenue'] + df['Shipping_Cost']
```

# **simpan file bersih**

```python
df.to_csv('cleaned_sales_data.csv', index=False)
```
