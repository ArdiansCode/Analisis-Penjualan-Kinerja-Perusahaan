# **Import Library**

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd
```

# **Muat Dataset**

```python
df = pd.read_csv("/content/cleaned_sales_data.csv")

df.head()
```

# **Buat Barchart untuk analisis lanjut**


```python
df_grouped = df.groupby('Product_Category')['Gross_Revenue'].sum().reset_index()

plt.bar(df_grouped['Product_Category'], df_grouped['Gross_Revenue'], color='skyblue')
plt.yscale('log')
plt.title('Grafik Penjualan Produk')
plt.xlabel('Category')
plt.ylabel('Gross_revenue')
plt.xticks(rotation=45)

def change_format(x, pos):
    return f'{x:,.0f}'.replace(',', '.')

formatter = ticker.FuncFormatter(change_format)
plt.gca().yaxis.set_major_formatter(formatter)

plt.tight_layout()
plt.show()
```

dari visualisasi data diatas dapat dilihat bahwa penjulan paling tinggi untuk triwulan pertama adalah barang yang berkategorikan Elekronik, sedangkan pada kategory lain bahkan tidak mencapai dari setengah penjualan barang elektronik, ini menandapakn bahwa lebih dari 60-70 persen pendapatan kita adalah dari category elekronik. untuk bisa menyeimbangkan penjualan kita, kita harus meningkatkan penjualan di kategory lain dengan cara buat promosi atau iklan.

# **Buat PieChart untuk analisis**

```python
Delivered_Status = len(df[df["Delivery_Status"] == "Delivered"])
Delayed_Status = len(df[df["Delivery_Status"] == "Delayed"])
Cancelled_Status = len(df[df["Delivery_Status"] == "Cancelled"])

labels = ["Delivered", "Delayed", "Cancelled"]
counting = [Delivered_Status, Delayed_Status, Cancelled_Status]

plt.pie(counting, labels=labels, autopct='%1.1f%%', startangle=90)
plt.axis('equal')
plt.title('digram penjualan')
plt.show()

```

dari grafik diatas dapat dilihat bahwa pada penjualawan pada triwulan pertama 35% penjualan kita terlihat kurang sehat, jumlah cancelled yang mencapai 18%, dandelayed 17% mendandakan penjualan kita kurang sehat, kita harus mengambil tindakan untuk pengiriman dan pengemasan barnag kepada pelanggan dipercepat.
